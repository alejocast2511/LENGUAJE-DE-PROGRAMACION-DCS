```python
import numpy as np
import matplotlib.pyplot as plt

def simular_cartera_montecarlo (retorno_anual, volatilidad_anual, correlacion_media, ponderaciones, dias_simulacion=252, num_simulaciones=10000):
  # 1. Parametros Iniciales y Conversion Diaria
  # dt es el incremento de tiempo, asumiendo retornos diarios.
  dt = 1 / dias_simulacion
  num_activos = len(ponderaciones)

  # Convertimos las tasas anuales a tasas diarias para el modelo MBG
  retorno_diario = retorno_anual / dias_simulacion
  volatilidad_diario = volatilidad_anual / np.sqrt(dias_simulacion)

  # 2. Creacion de la matriz de Covarianza

  covarianza_diaria = correlacion_media * volatilidad_diario **2
  matriz_cov =np.full((num_activos, num_activos), covarianza_diaria)
  np.fill_diagonal(matriz_cov, volatilidad_diario **2)

  # 3. Aplicacion descomposicion de cholesky

  try :
    L = NP.LINALG.CHOLESKY(matriz_cov)
  except np.linalg.LinAlgError:
    print("La matriz de covarianza no es definida positiva")
    return np.array([np.nan])

  rendimientos_finales =np.empty(num_simulaciones)

  # 4. Bucle de simulacion principal
  for i in range(num_simulaciones):
    z_no_corr = np.random.standard_normal((dias_simulacion, num_activos))

    z_corr = z_no_corr @ L.T

    drift = retorno_diario - 0.5 * volatilidad_diario**2
    choque_aleatorio = volatilidad_diario * z_corr

    retornos_activos_simulados = drift + choque_aleatorio

    retorno_cartera_diario = np.dot(retornos_activos_simulados, ponderaciones)

    rendimientos_final = np.prod(1 + retorno_cartera_diario) - 1
    rendimientos_finales[i] = rendimientos_final

  return rendimientos_finales



```
