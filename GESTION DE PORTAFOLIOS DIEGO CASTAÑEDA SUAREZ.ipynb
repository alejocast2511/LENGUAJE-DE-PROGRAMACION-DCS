#GESTION DE PORTAFOLIO DE IMPRESION

El presente trabajo simula la gestion de informacion sobre clientes de un fondo de inversion utilizando los conceptos repasados en clase.

##CONCEPTOS A TRABAJAR

  VARIABLES
  LISTAS
  TUPLAS
  DICCIONARIOS
  CICLOS
  CONDICIONALES

REALIZADO POR: DIEGO ALEJANDRO CASTAÑEDA SUAREZ.


```python
# INFORMACION DE CLIENTES

clientes = ["Juan Perez","Maria Rodriguez","Carlos Lopez","Ana Gomez"]
montos_invertidos = [10000, 15000, 20000, 18000]
rendimientos = [5.5, 6.0, 4.0, 7,5]

print(clientes)
print(montos_invertidos)
print(rendimientos)
```

```python
# TUPLAS

cliente_1 = (clientes[0], montos_invertidos[0], rendimientos[0])
cliente_2 = (clientes[1], montos_invertidos[1], rendimientos[1])
cliente_3 = (clientes[2], montos_invertidos[2], rendimientos[2])
cliente_4 = (clientes[3], montos_invertidos[3], rendimientos[3])

print("tuplas de clientes:")
print(cliente_1)
print(cliente_2)
print(cliente_3)
print(cliente_4)

```

```python
# DICCIONARIO PORTAFOLIO

portafolio = {
    "Juan Perez":{"monto_invertido":10000,"rendimiento":5.5},
    "Maria Rodriguez":{"monto_invertido":15000,"rendimiento":6.0}, "Carlos Lopez":{"monto_invertido":20000,"rendimiento":4.0},
    "Ana Gomez":{"monto_invertido":18000,"rendimiento":7.5}
    }

print("portafolio de clientes:")
print(portafolio)
```

```python
# EVALUACION DE RENDIMIENTO

print("clientes con rendimiento superior al 5%:")
for cliente, datos in portafolio.items():
    if datos["rendimiento"] > 5:
        print(cliente, "-", datos["rendimiento"], "%")
```

```python
#RENDIMIENTO TOTAL DEL PORTAFOLIO (EN DOLARES)

rendimiento_total = 0
for cliente, datos in portafolio.items():
    rendimiento_cliente = datos["monto_invertido"] * datos["rendimiento"] / 100
    rendimiento_total += rendimiento_cliente
    print("Rendimiento de", cliente, ":", rendimiento_cliente, "dolares")
print("Rendimiento total del portafolio:", rendimiento_total, "dolares")
```

```python
# CLIENTE CON MAYOR RENDIMIENTO

mayor_rendimiento = 0
cliente_top = ""
for cliente, datos in portafolio.items():
    rendimiento_cliente = datos["monto_invertido"] * datos["rendimiento"] / 100
    if rendimiento_cliente > mayor_rendimiento:
        mayor_rendimiento = rendimiento_cliente
        cliente_top = cliente
print("Cliente con mayor rendimiento absoluto:", cliente_top, "-", mayor_rendimiento, "dolares")
```

```python
# MODIFICACIONES DEL PORTAFOLIO

# 1. ACTUALIZACION DEL MONTO INVERTIDO
portafolio["Carlos Lopez" ]["monto_invertido"] = 25000
print("actualizacion del monto invertido de Carlos Lopez:", portafolio["Carlos Lopez"])


# 2. NUEVO CLIENTE
portafolio["Laura Fernandez"] = {"monto_invertido": 22000, "rendimiento": 6.8}
print("nuevo cliente ingresado:", portafolio["Laura Fernandez"])

# 3. REMOVER CLIENTE
# Add 'Ana Gomez' back to the dictionary before attempting to remove it
portafolio["Ana Gomez"] = {"monto_invertido": 18000, "rendimiento": 7.5}
portafolio.pop("Ana Gomez")
print("Portafolio despues de modificaciones:")
print(portafolio)
```
