## 2. ¿Qué es un Análisis de Portafolio de Inversión? 📚

Analizar un portafolio de inversion significa utilizar herramientas, teorias, bases, modelos en el proceso de toma de decisiones financieras con el objetivo final de optimizar la relacion que hay entre el rendimiento de una inversion, activos, bonos, etc.  y el riesgo que conlleva realizar esta inversion, es decir, maximizar el rendimiento y reducir el riesgo.




## 3. Investigación: Tipos de Análisis de Portafolio 🔍

### **Propuesta 1: Análisis de Rentabilidad y Riesgo**

El análisis de rentabilidad y riesgo se enfoca en el objetivo fundamental de todo inversionista: maximizar la rentabilidad esperada y minimizar el riesgo. La inversión se considera exitosa si logra el equilibrio óptimo entre estos dos elementos, junto con el factor tiempo.

1. **Rentabilidad (Retorno):** La rentabilidad de una inversión refleja el valor que se espera obtener en función del capital y el tipo de sector en el que se invierte. El rendimiento esperado de un portafolio se calcula como el promedio ponderado del rendimiento esperado de cada activo que lo compone.

2. **Riesgo (Volatilidad):** El riesgo se define como la variabilidad en los rendimientos futuros de una inversión en torno a su valor esperado. La posibilidad de que ocurra un evento desfavorable y los resultados difieran de los esperados siempre está presente debido a la incertidumbre.

3. **Bases Académicas y Cuantificación:** La principal herramienta teórica para abordar estos dos componentes es el Modelo Media-Varianza (MMV) de Harry Markowitz, que utiliza el valor esperado (media) para el rendimiento y la Varianza o Desviación Estándar (${\sigma}$) para medir el riesgo o volatilidad. Otros modelos, como el CAPM (Capital Asset Pricing Model) de William Sharpe, explican la relación entre el rendimiento requerido de un activo y su riesgo sistemático.

4. **Métricas Ajustadas por Riesgo:** El rendimiento no debe verse de forma aislada. Para evaluar la eficiencia con la que se logró un rendimiento dado el riesgo asumido, se utilizan métricas como el Índice de Sharpe, que cuantifica el exceso de rentabilidad por unidad de riesgo. Otras métricas ajustadas por riesgo son el Índice de Sortino y el Índice de Treynor. El Valor en Riesgo (VaR) es una técnica estadística que cuantifica la máxima pérdida esperada en un intervalo de tiempo determinado y nivel de confianza dado

### **Propuesta 2:  Diversificación y Correlación**

La diversificación es un principio fundamental de la inversión, que busca mitigar el riesgo y mejorar el potencial de rendimiento positivo.
1. **Diversificación:** Consiste en invertir en más de un activo para reducir el nivel de riesgo asociado con factores específicos de una compañía. Actúa como una red de seguridad financiera que protege el patrimonio de la volatilidad.

2. **Gestión del Riesgo:** Al diversificar, las ganancias de unos activos pueden compensar las pérdidas de otros, dado que el precio de todos los activos no cambia en proporciones iguales ni debido a los mismos factores. La diversificación es la herramienta clave para gestionar el riesgo no sistemático o específico (inherente al emisor). Sin embargo, la diversificación no elimina el riesgo sistemático (de mercado), aunque este puede combatirse mediante la diversificación internacional.

3. **Correlación y Eficiencia:** El riesgo de un portafolio depende principalmente del coeficiente de correlación entre sus instrumentos. Para que la diversificación sea efectiva, se deben buscar activos cuya covarianza o correlación sea baja. Si la correlación entre los activos es perfectamente negativa (${\rho}_{AB} = -1$), la diversificación puede hacer desaparecer completamente el riesgo.

4. **Tipos de Diversificación:** La diversificación se extiende a través de clases de activos (acciones, bonos, etc.), regiones geográficas, sectores, industrias (tecnología, salud, energía) y tamaños de empresas (gran, mediana y pequeña capitalización)


### **Propuesta 3: Simulación de Escenarios (Conservador, Moderado, Agresivo)**  

La simulación de escenarios, en el contexto de la gestión de carteras, implica definir el perfil del inversionista y, en base a este, construir portafolios que se ajusten a diferentes tolerancias al riesgo.

1. **Base del Perfil:** El tipo de portafolio se determina por factores como la edad, el estatus financiero, los objetivos, las metas, el horizonte temporal y el grado de aversión al riesgo del individuo.

2. **Tipos de Portafolio por Perfil (Simulación de Escenarios):**

    2.1  *Portafolio Conservador:* Está diseñado para el inversor cauto y adverso al riesgo, con el objetivo principal de preservar el capital. Los activos seleccionados tienen un riesgo mínimo y rendimientos mínimos. La participación en renta variable suele ser menor al 20%, predominando la deuda gubernamental o corporativa de mediano y largo plazo.

    2.2 *Portafolio Moderado:* El inversionista busca un equilibrio entre mitigar el riesgo y obtener rendimientos atractivos. Utiliza la diversificación en la mayor medida posible y combina activos de riesgo medio y bajo con una menor proporción de activos de alto riesgo. La participación en renta variable suele ser entre 20% y 60%.

    2.3 *Portafolio Agresivo (o Experto):* El inversionista está dispuesto a asumir una elevada cantidad de riesgo a cambio de la posibilidad de hacer crecer y multiplicar el capital. Estas estrategias suelen tener un horizonte temporal mayor y una mayor participación de renta variable e instrumentos derivados.

3. **Herramientas Avanzadas para Simulación:** Además de la clasificación por perfil, la gestión del riesgo se complementa con herramientas de simulación. La Simulación Monte Carlo es una metodología que permite modelar escenarios diversos y generar una gran cantidad de combinaciones de activos para identificar el conjunto óptimo. Asimismo, la técnica de Valor en Riesgo (VaR) es crucial, ya que permite a los gestores de riesgo conocer cuál podría ser la mayor pérdida esperada en un escenario adverso (el "peor escenario posible").
